```python
from pathlib import Path

import pandas as pd
output_dir = "../visualizations"
gemini3_scores = "../data_dir/checkpoints/gemini30preview/gemini30preview_scores_exclude_small_samples.json"

gpt51_scores = "../data_dir/checkpoints/gpt-5.1/gpt-5.1_scores_exclude_small_samples.json"

gemini31_scores="../data_dir/checkpoints/gemini31propreview/gemini31propreview_scores_exclude_small_samples.json"




gemini2_scores = "../data_dir/checkpoints/gemini25pro/gemini25pro_scores_exclude_small_samples.json"
```

```python

pd_gemini_3 = pd.read_json(gemini3_scores)
pd_gemini_3["PA"] = pd_gemini_3["PA"].apply(lambda x: round(x, 2) if isinstance(x, float) else x)
pd_gemini_3 = pd_gemini_3.set_index(["name", "PA", "S", "m"])

pd_gemini_3.head()
```

```python

pd_gemini_2 = pd.read_json(gemini2_scores)
pd_gemini_2["PA"] = pd_gemini_2["PA"].apply(lambda x: round(x, 2) if isinstance(x, float) else x)
pd_gemini_2  = pd_gemini_2.set_index(["name", "PA", "S", "m"])

pd_gemini_2.head()
```

```python

pd_gpt_51 = pd.read_json(gpt51_scores)
pd_gpt_51["PA"] = pd_gpt_51["PA"].apply(lambda x: round(x, 2) if isinstance(x, float) else x)
pd_gpt_51  = pd_gpt_51.set_index(["name", "PA", "S", "m"])

pd_gpt_51.head()
```

```python


pd_gemini_31 = pd.read_json(gemini31_scores)
pd_gemini_31["PA"] = pd_gemini_31["PA"].apply(lambda x: round(x, 2) if isinstance(x, float) else x)
pd_gemini_31  = pd_gemini_31.set_index(["name", "PA", "S", "m"])

pd_gemini_31.head()
```

```python

def df_compare_scores( df_target_model, df_base_model ):
    df_compare_scores = pd.DataFrame()
    df_compare_scores["base_f1-score"]=df_base_model["model_f1-score"]
    df_compare_scores["base_accuracy"]=df_base_model["model_accuracy"]

    df_compare_scores["target_f1-score"]=df_target_model["model_f1-score"]
    df_compare_scores["target_accuracy"]=df_target_model["model_accuracy"]

    df_compare_scores["delta_target_vs_base_model_f1-score"]= df_compare_scores["target_f1-score"] - df_compare_scores["base_f1-score"]
    df_compare_scores["delta_target_vs_base_model_accuracy"]= df_compare_scores["target_accuracy"]-df_compare_scores["base_accuracy"]

    return df_compare_scores

scores_df = df_compare_scores(pd_gemini_2, pd_gemini_3)



# F1

scores_df.describe()
```
