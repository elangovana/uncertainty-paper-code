```python
import pandas as pd

gemini3_scores = "../data_dir/checkpoints/gemini30preview/gemini30preview_scores_exclude_small_samples.json"

gemini2_scores = "../data_dir/checkpoints/gemini25pro/gemini25pro_scores_exclude_small_samples.json"
```

```python

pd_gemini_3 = pd.read_json(gemini3_scores)
pd_gemini_3["PA"] = pd_gemini_3["PA"].apply(lambda x: round(x, 2) if isinstance(x, float) else x)
pd_gemini_3 = pd_gemini_3.set_index(["name", "PA", "S", "m"])

pd_gemini_3.head()
```

```python

pd_gemini_2 = pd.read_json(gemini2_scores)
pd_gemini_2["PA"] = pd_gemini_2["PA"].apply(lambda x: round(x, 2) if isinstance(x, float) else x)
pd_gemini_2  = pd_gemini_2.set_index(["name", "PA", "S", "m"])

pd_gemini_2.head()
```

```python
df_compare_scores = pd.DataFrame()
df_compare_scores["gemini2_f1-score"]=pd_gemini_2["model_f1-score"]
df_compare_scores["gemini2_accuracy"]=pd_gemini_2["model_accuracy"]
df_compare_scores["gemini3_f1-score"]=pd_gemini_3["model_f1-score"]
df_compare_scores["gemini3_accuracy"]=pd_gemini_3["model_accuracy"]
df_compare_scores["delta_gemini_3_vs_2_accuracy"]=df_compare_scores["gemini3_accuracy"] -df_compare_scores["gemini2_accuracy"]
df_compare_scores["delta_gemini_3_vs_2_f1-score"]=df_compare_scores["gemini3_f1-score"] -df_compare_scores["gemini2_f1-score"]
df_compare_scores.head()
```
